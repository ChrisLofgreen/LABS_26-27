### 6.
listings[['name', 'price']].sort_values('price', ascending=False).head(3) shows 999.60, 999.00 and 999.00.
This is because the datatype is string and the list is decending (highest first). The first character is "$" which should be the same for all.
Then comes the second character where "9" is highest and so on until we come to the deciding number for this list which is "6" (as in .60).
In short they are not sorted by numeric value but by the value of each character in the string, so "1,000" is concidered lower than "999".

```python
print(listings["price"].head(3))
print(listings["price"].dtypes)
```

```python
listings[['name', 'price']].sort_values('price', ascending=False).head(3)
```

## Part 2: Go Deeper

### 1.
2331 entries are in central which is about 51%.
The share of private rooms in central is about 15.1% and about 19.4% outside of central (17.2% overall)

```python
central = ["Södermalms", "Östermalms", "Norrmalms", "Kungsholmens"]
mask = listings["neighbourhood_cleansed"].isin(central)

print(mask.sum())
print(mask.mean())
```

```python
private_room_central = listings[mask]["room_type"] == "Private room"
private_room_not_central = listings[~mask]["room_type"] == "Private room"
private_room_overall = listings["room_type"] == "Private room"

print(private_room_central.mean())
print(private_room_not_central.mean())
print(private_room_overall.mean())
```

### 2.
About 55.5% of "Entire home/apt" have accommodation for 4 or more people, for the whole dataset that number is ca 9.3%.
This tells us that rentals listed as "Entire home/apt" are (in general) more likley to have accomotation for 4 or more people compared to the whole dataset.
It does NOT tell us, that this category is the one where we are MOST likley to find 4+ accomodation (among other things).


```python
home_apt = listings["room_type"] == "Entire home/apt"

home_apt_four = listings[home_apt]["accommodates"] >= 4
four_not_home_apt = listings[~home_apt]["accommodates"] >= 4

print(home_apt_four.sum())
print(home_apt_four.mean())
print(four_not_home_apt.mean())
```
