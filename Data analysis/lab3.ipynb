```python
import pandas as pd

loans_raw = pd.read_csv("data/loans.csv")

loans_clean = loans_raw.copy()

# Clean strings in "genre"
loans_clean["genre"] = loans_clean["genre"].str.strip().str.lower()

loans_clean["genre"] = (loans_clean["genre"]
                  .replace("children's", "children")
                  .replace("sci-fi", "science fiction")
                  .replace("sci fi", "science fiction")
                  .str.title()
                  )

# Make "returned" booleans
loans_clean["returned"] = loans_clean["returned"].str.strip().str.lower().map({"yes": True, "y": True, "n": False, "no": False})

# Clean "late_fee" strings
late_fee_fixed = loans_clean["late_fee"].str.strip()
late_fee_fixed = late_fee_fixed.str.replace(" ", "")
late_fee_fixed = late_fee_fixed.str.replace("kr", "")
late_fee_fixed = late_fee_fixed.str.replace(",", ".")

# Make "late_fee" floats
loans_clean["late_fee"] = late_fee_fixed.astype(float)

# Make "loan_date" into datetime objects
new_return_dates = pd.to_datetime(loans_clean["return_date"], errors="coerce")
new_dates = pd.to_datetime(loans_clean["loan_date"], errors="coerce")
loans_clean["loan_date"] = new_dates
loans_clean["return_date"] = new_return_dates

# Change return_dates that are before the loan_dates to equal loan_date
mask = loans_clean["return_date"] < loans_clean["loan_date"]
loans_clean.loc[mask, "return_date"] = loans_clean.loc[mask, "loan_date"]

# Drop duplicates and reset index
loans_clean = loans_clean.drop_duplicates().reset_index(drop=True)

# Drop empty column "notes"
loans_clean = loans_clean.drop(columns="notes")
```
