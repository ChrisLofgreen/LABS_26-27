### 2.

- There are 28 genres before cleanup
- 9 after strip() and lower()

```python
loans["genre"].unique()
```

```python
loans["genre"].value_counts()
```

```python
print(loans["genre"].nunique())
print(loans["genre"].str.strip().nunique())
print(loans["genre"].str.strip().str.lower().nunique())
```

```python
print(loans["genre"].str.strip().str.lower().value_counts())
```

```python
loans["genre"] = loans["genre"].str.strip().str.lower()

print(loans["genre"].value_counts())
```

```python
print((loans["genre"]
                  .replace("children's", "children")
                  .replace("sci-fi", "science fiction")
                  .replace("sci fi", "science fiction")
                  .str.title()
                  ).value_counts())
```

```python
loans["genre"] = (loans["genre"]
                  .replace("children's", "children")
                  .replace("sci-fi", "science fiction")
                  .replace("sci fi", "science fiction")
                  .str.title()
                  )
```

```python
loans["returned"] = loans["returned"].str.strip().str.lower().map({"yes": True, "y": True, "n": False, "no": False})

print(loans["returned"].unique())
```
